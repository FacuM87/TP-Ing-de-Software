## 1. Problema de Negocio

### Contexto
Nos situamos en una empresa argentina de escala regional dedicada a la producción y exportación de manufacturas agroindustriales e industriales ("AgroSur Exportaciones S.A."), con presencia comercial activa en el Mercosur y en el resto del continente americano (principalmente Brasil, Chile, Paraguay, Colombia, México y Estados Unidos).

### Planteo del Problema
La demanda de exportaciones de la empresa, la colocación de stocks, la fijación de precios y los planes de producción dependen fuertemente del pulso y del ciclo económico de los países destinatarios, así como de la evolución de la economía local en Argentina:

* Actualmente, las decisiones de abastecimiento, planificación de la producción y asignación de presupuestos comerciales se realizan de manera **reactiva**, apoyándose en balances pasados o en noticias periodísticas de coyuntura.
* Cuando un mercado de exportación relevante entra en desaceleración o contracción económica, la empresa se entera con varios meses de rezago. Esto genera **sobrestock acumulado**, mayores costos de almacenamiento logístico y la necesidad de liquidar mercadería a menor precio.
* A la inversa, cuando una economía vecina entra en una fase expansiva de aceleración del consumo, la empresa no cuenta con la previsión suficiente para incrementar turnos de producción ni asegurar cupos logísticos, perdiendo participación de mercado frente a competidores extranjeros.

### Solución propuesta con Minería de Datos
Implementar un sistema de análisis, monitoreo y pronóstico del índice de actividad económica regional utilizando técnicas de Ciencia de Datos. El sistema permitirá detectar de forma temprana fases expansivas o contractivas, estudiar el co-movimiento y correlación entre Argentina y las demás economías de América, y anticipar el comportamiento macroeconómico para respaldar la toma de decisiones comerciales y operativas.

## 2. Alternativa: ¿Cómo construir el set de datos si no existiera?

Si este repositorio unificado no estuviese disponible, la empresa debería implementar un proceso propio de recolección y homologación de datos (*ETL - Extract, Transform, Load*):

3. **Re-expresión a base homogénea:** Cada país publica su índice con un año base diferente (ej. base 2004=100, base 2010=100, etc.). Sería necesario re-indexar todas las series a una base de referencia común para permitir su comparación y análisis conjunto.
4. **Desestacionalización metodológica:** En los países donde solo se disponga de la serie bruta, aplicar algoritmos estandarizados como **X-13ARIMA-SEATS** o **STL Decomposition** para extraer la componente desestacionalizada.
5. **Consolidación en un Data Warehouse / Data Lake:** Unificar todas las series en una tabla maestra relacional indexada por la clave temporal (`Fecha`).

## 3. Desafíos y Problemas de Calidad de los Datos Detectados

Durante la exploración inicial se identifican los siguientes aspectos de calidad que impactan en el análisis:

   * Países como Argentina, Brasil, Paraguay, Costa Rica, México, Guatemala, Canadá, Estados Unidos y Honduras cuentan con el **100% de los datos** (229 registros mensuales completos desde 2004 a 2023).

# FASE 4: Modelado (Modeling)

En esta fase se define la variable objetivo, se seleccionan los algoritmos y se calibran sus hiperparámetros.

---

## 1. Variable Objetivo

* **Variable objetivo (Y):** `Nivel_Actividad`, construida a partir del índice de Argentina (serie desestacionalizada) y su mediana histórica:
  * **`Alta Actividad`** (≥ 143,10): períodos de expansión, con mayor demanda interna y operativa.
  * **`Baja Actividad`** (< 143,10): períodos de desaceleración, estancamiento o contracción.
* **Justificación de negocio:** anticipar en qué régimen operará la economía permite activar a tiempo decisiones concretas: aumentar turnos y compras de insumos, o resguardar liquidez y moderar inventarios.
* **Variables predictoras (X):** `Brasil`, `EstadosUnidos`, `Mexico` y `Paraguay`.

---

## 2. Algoritmos Candidatos

| Algoritmo | Ventajas | Limitaciones | Decisión |
| :--- | :--- | :--- | :--- |
| **Regresión Logística** | Simple y rápida. Devuelve probabilidades y coeficientes que indican cómo influye cada variable. | Supone una frontera de decisión lineal y sus coeficientes se distorsionan con multicolinealidad. | **Seleccionado** como modelo base interpretable. |
| **Random Forest** | Captura relaciones no lineales, es robusto a la multicolinealidad y no requiere escalar las variables. Indica la importancia de cada variable. | Menos transparente que un modelo lineal o un árbol individual. | **Seleccionado** como modelo de mayor capacidad predictiva. |
| **Árbol de Decisión individual** | Reglas explícitas y fáciles de leer. | Alta varianza: pequeños cambios en los datos alteran los cortes y tiende al sobreajuste. | Descartado: Random Forest conserva su lógica y corrige su inestabilidad. |
| **k-Vecinos más Cercanos (k-NN)** | Predice por similitud con meses históricos. | Sensible a la escala y no indica qué variables explican la predicción. | Descartado por baja explicabilidad. |
| **Máquinas de Vectores de Soporte (SVM)** | Eficaz con fronteras complejas. | Difícil de interpretar y sensible a la elección del kernel. | Descartado por baja explicabilidad. |

---

## 3. Algoritmos Seleccionados

### Regresión Logística
* **Fundamento:** estima la probabilidad de que un mes sea de *Alta Actividad* aplicando la función sigmoide a una combinación lineal de las predictoras. El parámetro **C** controla la regularización: valores bajos simplifican el modelo y valores altos lo ajustan más a los datos.
* **Utilidad para el negocio:** el signo y la magnitud de cada coeficiente muestran qué economías empujan hacia un régimen u otro, y la probabilidad estimada funciona como nivel de confianza de la alerta.

### Random Forest
* **Fundamento:** entrena muchos árboles de decisión (300 en este caso), cada uno sobre una muestra aleatoria de los meses y considerando subconjuntos aleatorios de variables en cada corte. La predicción final es el voto mayoritario de los árboles. Se calibra la **profundidad máxima** de cada árbol.
* **Utilidad para el negocio:** al promediar muchos árboles, ofrece predicciones más estables que un árbol individual, y su ranking de **importancia de variables** indica qué mercados conviene monitorear con más atención.

### Estrategia de calibración
Los hiperparámetros (C y profundidad máxima) se eligen por **validación cruzada de 5 particiones sobre el conjunto de entrenamiento**. El conjunto de prueba queda reservado y se usa una sola vez, en la Fase 5, para obtener una medida de rendimiento no sesgada.

```python
# FASE 4: Carga de librerías para Modelado

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
print("✓ Librerías para Modelado cargadas exitosamente.")
```

```python
# FASE 4: Modelado - Partición de datos y calibración de Regresión Logística y Random Forest

# 1. Matriz de predictoras (X) y variable objetivo (y), construida en la Fase 3
variables_predictoras = ['Brasil', 'EstadosUnidos', 'Mexico', 'Paraguay']
X = df_estudio[variables_predictoras]
y = df_estudio['Nivel_Actividad']

# 2. Partición Train / Test (80% / 20%), estratificada y con semilla fija para reproducibilidad
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
print(f"✓ Conjunto de Entrenamiento: {X_train.shape[0]} meses")
print(f"✓ Conjunto de Prueba:        {X_test.shape[0]} meses (reservado para la Fase 5)")

# 3. Calibración de hiperparámetros con validación cruzada de 5 particiones sobre el entrenamiento.
#    El conjunto de prueba no interviene en la elección: se usa una sola vez, en la evaluación final.
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Modelo 1: Regresión Logística, con estandarización previa (escalas muy distintas entre países).
#           C controla la regularización: valores bajos = modelo más simple.
valores_C = [0.01, 0.1, 1, 10, 100]
cv_logistica = [cross_val_score(make_pipeline(StandardScaler(), LogisticRegression(C=c, max_iter=1000)),
                                X_train, y_train, cv=cv) for c in valores_C]

# Modelo 2: Random Forest (300 árboles); se calibra la profundidad máxima de cada árbol.
profundidades = [2, 3, 4, 5, None]
cv_rf = [cross_val_score(RandomForestClassifier(n_estimators=300, max_depth=p, random_state=42),
                         X_train, y_train, cv=cv) for p in profundidades]

mejor_C = valores_C[int(np.argmax([s.mean() for s in cv_logistica]))]
mejor_prof = profundidades[int(np.argmax([s.mean() for s in cv_rf]))]

tabla_cv = pd.DataFrame({
    'Modelo': ['Regresión Logística'] * len(valores_C) + ['Random Forest'] * len(profundidades),
    'Hiperparámetro': [f"C = {c}" for c in valores_C] + [f"max_depth = {p if p else 'sin límite'}" for p in profundidades],
    'Exactitud CV (media)': [f"{s.mean()*100:.2f}%" for s in cv_logistica + cv_rf],
    'Desvío CV': [f"{s.std()*100:.2f}%" for s in cv_logistica + cv_rf]
})
print("\n" + "="*80)
print("CALIBRACIÓN DE HIPERPARÁMETROS (VALIDACIÓN CRUZADA 5-FOLD SOBRE ENTRENAMIENTO)")
print("="*80)
display(tabla_cv)
print(f"\n★ Mejor Regresión Logística: C = {mejor_C}")
print(f"★ Mejor Random Forest:       max_depth = {mejor_prof if mejor_prof else 'sin límite'}")

# Curvas de calibración
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 4.5))
ax1.errorbar(range(len(valores_C)), [s.mean()*100 for s in cv_logistica], yerr=[s.std()*100 for s in cv_logistica],
             marker='o', capsize=4, color='#1f77b4')
ax1.set_xticks(range(len(valores_C)))
ax1.set_xticklabels([str(c) for c in valores_C])
ax1.set_title("Regresión Logística: Exactitud CV según C", fontweight='bold')
ax1.set_xlabel("Parámetro de regularización C")
ax1.set_ylabel("Exactitud CV (%)")

ax2.errorbar(range(len(profundidades)), [s.mean()*100 for s in cv_rf], yerr=[s.std()*100 for s in cv_rf],
             marker='o', capsize=4, color='#2ca02c')
ax2.set_xticks(range(len(profundidades)))
ax2.set_xticklabels([str(p) if p else 'sin límite' for p in profundidades])
ax2.set_title("Random Forest: Exactitud CV según profundidad", fontweight='bold')
ax2.set_xlabel("Profundidad máxima de cada árbol (max_depth)")
ax2.set_ylabel("Exactitud CV (%)")
plt.tight_layout()
plt.show()

# 4. Entrenamiento de los modelos definitivos sobre todo el conjunto de entrenamiento
modelo_logistica = make_pipeline(StandardScaler(), LogisticRegression(C=mejor_C, max_iter=1000))
modelo_logistica.fit(X_train, y_train)

modelo_rf = RandomForestClassifier(n_estimators=300, max_depth=mejor_prof, random_state=42)
modelo_rf.fit(X_train, y_train)

# 5. Interpretación de los modelos
# Coeficientes de la logística expresados a favor de "Alta Actividad" (sobre variables estandarizadas)
clases_log = list(modelo_logistica.classes_)
signo = 1 if clases_log[1] == 'Alta Actividad' else -1
coeficientes = pd.Series(signo * modelo_logistica[-1].coef_[0], index=variables_predictoras).sort_values()
importancias = pd.Series(modelo_rf.feature_importances_, index=variables_predictoras).sort_values()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 4))
ax1.barh(coeficientes.index, coeficientes.values,
         color=['#2ca02c' if v > 0 else '#d62728' for v in coeficientes.values])
ax1.axvline(0, color='gray', linewidth=1)
ax1.set_title("Regresión Logística: Coeficientes\n(positivo = empuja hacia Alta Actividad)", fontweight='bold')
ax1.set_xlabel("Coeficiente (variables estandarizadas)")

ax2.barh(importancias.index, importancias.values * 100, color=[colores[p] for p in importancias.index])
ax2.set_title("Random Forest: Importancia de cada Variable", fontweight='bold')
ax2.set_xlabel("Importancia (%)")
plt.tight_layout()
plt.show()

print("Coeficientes de la Regresión Logística (a favor de Alta Actividad):")
display(coeficientes.sort_values(ascending=False).round(2).to_frame('Coeficiente'))
print("Importancia de variables en Random Forest:")
display((importancias.sort_values(ascending=False) * 100).round(1).to_frame('Importancia (%)'))
```

   * **Random Forest:** mejor configuración con **max_depth = 4** (300 árboles), con una exactitud en validación cruzada de **93,48% ± 4,02%**. Árboles más profundos no mejoran el resultado, lo que indica que limitar la profundidad evita el sobreajuste.

Este resultado confirma que, con predictoras tan correlacionadas, los coeficientes de la logística deben leerse con cautela.

# FASE 5: Evaluación de los Modelos (Evaluation)

En esta etapa evaluamos el rendimiento de los modelos implementados sobre el conjunto de datos de prueba (*Test Set*), el cual contiene datos históricos que los algoritmos nunca observaron durante la etapa de entrenamiento. Esto permite medir la capacidad de generalización (complementada con una validación temporal) y asegurar que la solución responde a los objetivos comerciales planteados en la Fase 1.

## 1. Investigación y Definición de Métricas de Evaluación

Para medir el rendimiento de los modelos de clasificación supervisada se utilizan las siguientes métricas estadísticas:

### 1.1 Matriz de Confusión
Es una tabla de doble entrada que compara los valores reales de la muestra contra las predicciones emitidas por el modelo. Permite descomponer los resultados en cuatro categorías:

* **Verdaderos Positivos (VP):** El modelo predijo correctamente un período de *Alta Actividad*.
* **Verdaderos Negativos (VN):** El modelo predijo correctamente un período de *Baja Actividad*.
* **Falsos Positivos (FP) - Error Tipo I:** El modelo predijo *Alta Actividad*, pero en la realidad el país atravesó *Baja Actividad*.
  * *Impacto en el negocio:* Es un error costoso porque la empresa incrementa innecesariamente sus turnos de producción y compras de materias primas, acumulando sobrestock no vendido que encarece los costos financieros y logísticos.
* **Falsos Negativos (FN) - Error Tipo II:** El modelo predijo *Baja Actividad*, pero la economía experimentó una fase de *Alta Actividad*.
  * *Impacto en el negocio:* Se pierde costo de oportunidad; la empresa no alcanza a satisfacer los pedidos por falta de stock y los clientes recurren a competidores.

### 1.2 Exactitud (Accuracy)
Representa el porcentaje general de clasificaciones correctas sobre el total de casos evaluados:
$$\text{Accuracy} = \frac{\text{VP} + \text{VN}}{\text{VP} + \text{VN} + \text{FP} + \text{FN}}$$

### 1.3 Precisión (Precision)
Mide la confiabilidad de las alertas de crecimiento: de todos los meses que el modelo catalogó como "Alta Actividad", ¿cuántos lo fueron verdaderamente?
$$\text{Precision} = \frac{\text{VP}}{\text{VP} + \text{FP}}$$

### 1.4 Sensibilidad / Cobertura (Recall)
Mide la capacidad de captura del modelo: de todos los meses que realmente fueron de "Alta Actividad", ¿qué proporción logró detectar el modelo?
$$\text{Recall} = \frac{\text{VP}}{\text{VP} + \text{FN}}$$

### 1.5 F1-Score
Es la media armónica entre Precisión y Recall. Proporciona una medida balanceada que no se distorsiona fácilmente ante posibles desbalances en los datos:
$$\text{F1-Score} = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$$

```python
# FASE 5: Carga de librerías para Evaluación

from sklearn.base import clone
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
print("✓ Librerías para Evaluación cargadas exitosamente.")
```

```python
# FASE 5: Evaluación Formal, Comparación de Modelos y Matrices de Confusión

modelos = {'Regresión Logística': modelo_logistica, 'Random Forest': modelo_rf}
hiperparametros = {'Regresión Logística': f"C = {mejor_C}",
                   'Random Forest': f"max_depth = {mejor_prof if mejor_prof else 'sin límite'}"}
exactitud_cv = {'Regresión Logística': max(s.mean() for s in cv_logistica),
                'Random Forest': max(s.mean() for s in cv_rf)}
etiquetas = ['Alta Actividad', 'Baja Actividad']  # clase positiva: Alta Actividad

# 1. Evaluación sobre el conjunto de prueba (46 meses nunca vistos durante la calibración)
filas, predicciones = [], {}
for nombre, modelo in modelos.items():
    y_pred = modelo.predict(X_test)
    predicciones[nombre] = y_pred
    filas.append({
        'Modelo': nombre,
        'Hiperparámetro': hiperparametros[nombre],
        'Exactitud CV': f"{exactitud_cv[nombre]*100:.2f}%",
        'Exactitud Test': f"{accuracy_score(y_test, y_pred)*100:.2f}%",
        'Precisión (Alta)': f"{precision_score(y_test, y_pred, pos_label='Alta Actividad')*100:.2f}%",
        'Recall (Alta)': f"{recall_score(y_test, y_pred, pos_label='Alta Actividad')*100:.2f}%",
        'F1-Score (Alta)': f"{f1_score(y_test, y_pred, pos_label='Alta Actividad')*100:.2f}%"
    })

print("="*80)
print("TABLA COMPARATIVA DE RENDIMIENTO SOBRE EL CONJUNTO DE PRUEBA")
print("="*80)
display(pd.DataFrame(filas))

for nombre in modelos:
    print(f"\n--- Informe de clasificación: {nombre} ---")
    print(classification_report(y_test, predicciones[nombre], digits=3))

# Matrices de confusión
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
for i, (nombre, cmap) in enumerate(zip(modelos, ['Blues', 'Greens'])):
    cm = confusion_matrix(y_test, predicciones[nombre], labels=etiquetas)
    sns.heatmap(cm, annot=True, fmt='d', cmap=cmap, cbar=False, xticklabels=etiquetas, yticklabels=etiquetas,
                annot_kws={'size': 14, 'weight': 'bold'}, ax=ax[i])
    ax[i].set_title(f"Matriz de Confusión: {nombre}\nExactitud: {accuracy_score(y_test, predicciones[nombre])*100:.2f}%",
                    fontweight='bold')
    ax[i].set_xlabel("Predicción del Modelo", fontweight='bold')
    ax[i].set_ylabel("Valor Real", fontweight='bold')
plt.tight_layout()
plt.show()

# 2. Validación temporal: entrenar con 2004-2018 y evaluar sobre 2019-2023.
#    Simula el uso real del sistema: predecir períodos que el modelo nunca vio.
corte = df_estudio['Fecha'] < '2019-01-01'
X_pasado, y_pasado = X[corte], y[corte]
X_futuro, y_futuro = X[~corte], y[~corte]

filas_temp = []
for nombre, modelo in modelos.items():
    modelo_temp = clone(modelo).fit(X_pasado, y_pasado)
    y_pred_temp = modelo_temp.predict(X_futuro)
    cm = confusion_matrix(y_futuro, y_pred_temp, labels=etiquetas)
    filas_temp.append({
        'Modelo': nombre,
        'Exactitud (2019-2023)': f"{accuracy_score(y_futuro, y_pred_temp)*100:.2f}%",
        'Falsos Positivos': cm[1, 0],
        'Falsos Negativos': cm[0, 1]
    })

print("\n" + "="*80)
print(f"VALIDACIÓN TEMPORAL: entrenamiento 2004-2018 ({len(X_pasado)} meses) / prueba 2019-2023 ({len(X_futuro)} meses)")
print("="*80)
display(pd.DataFrame(filas_temp))
```

```python
# FASE 6: Simulador de Escenarios embebido en el notebook (versión reducida de la Vista 3)

import ipywidgets as widgets
from IPython.display import display, HTML

# Valores del último mes disponible: punto de partida del simulador
valores_iniciales = {p: round(float(df_estudio[p].iloc[-1]), 1) for p in variables_predictoras}
ultimo_mes = df_estudio['Fecha'].iloc[-1].strftime('%m/%Y')

def crear_slider(pais, descripcion, paso):
    """Slider limitado al rango histórico del país; valor inicial = último mes disponible."""
    return widgets.FloatSlider(value=valores_iniciales[pais],
                               min=round(float(df_estudio[pais].min()), 1),
                               max=round(float(df_estudio[pais].max()), 1), step=paso,
                               description=descripcion, continuous_update=False,
                               style={'description_width': '160px'}, layout=widgets.Layout(width='550px'))

sliders = {
    'Brasil': crear_slider('Brasil', 'Brasil (IBC-Br):', 0.5),
    'EstadosUnidos': crear_slider('EstadosUnidos', 'EE.UU. (PIB, miles de M USD):', 50),
    'Mexico': crear_slider('Mexico', 'México (IGAE):', 0.5),
    'Paraguay': crear_slider('Paraguay', 'Paraguay (IMAEP):', 0.5)
}
boton_simular = widgets.Button(description='Simular escenario', button_style='primary', icon='calculator',
                               layout=widgets.Layout(width='220px', height='40px'))
boton_reset = widgets.Button(description=f'Volver a {ultimo_mes}', icon='undo',
                             layout=widgets.Layout(width='220px', height='40px'))

def tarjeta_resultado():
    """Calcula la predicción para los valores actuales de los sliders y arma la tarjeta HTML."""
    valores = {p: s.value for p, s in sliders.items()}
    escenario = pd.DataFrame([valores])[variables_predictoras]
    prob_alta = modelo_rf.predict_proba(escenario)[0][list(modelo_rf.classes_).index('Alta Actividad')] * 100

    # Semáforo con los umbrales definidos en la Fase 6 (criterio de uso de la Fase 5)
    if prob_alta >= 70:
        color, fondo, estado = '#28a745', '#e8f5e9', '🟢 ALTA ACTIVIDAD'
        recomendacion = ("Escenario favorable: evaluar aumento de turnos, anticipar compras de insumos "
                         "y asegurar cupos logísticos, previa validación de los analistas.")
    elif prob_alta > 30:
        color, fondo, estado = '#e0a800', '#fff8e1', '🟡 ZONA DE INCERTIDUMBRE'
        recomendacion = ("Señal no concluyente: mantener niveles actuales de producción y stock "
                         "y monitorear los próximos datos publicados.")
    else:
        color, fondo, estado = '#dc3545', '#ffebee', '🔴 BAJA ACTIVIDAD'
        recomendacion = ("Escenario desfavorable: moderar la acumulación de stock, controlar costos fijos "
                         "y orientar esfuerzos comerciales a los mercados más estables.")

    tipo = (f"Último mes disponible ({ultimo_mes})" if valores == valores_iniciales else "Escenario simulado")
    detalle = " · ".join(f"{p}: {v:,.1f}" for p, v in valores.items())
    return f"""
    <div style="background-color:{fondo}; border-left:6px solid {color}; padding:16px; border-radius:6px;
                margin-top:12px; font-family:sans-serif; color:#222;">
        <p style="font-size:13px; margin:0 0 6px 0;"><b>{tipo}</b> — {detalle}</p>
        <h3 style="color:{color}; margin:6px 0;">{estado}</h3>
        <p style="font-size:15px;"><b>Probabilidad de Alta Actividad en Argentina:</b> {prob_alta:.1f}%</p>
        <p style="font-size:14px;"><b>Recomendación:</b> {recomendacion}</p>
        <p style="font-size:12px; color:#555;">Modelo: Random Forest (exactitud en test {accuracy_score(y_test, modelo_rf.predict(X_test))*100:.1f}%).
        Uso orientado al análisis de escenarios dentro de rangos históricos; no constituye un pronóstico de períodos futuros (ver Fase 5).</p>
    </div>"""

def simular(_):
    tarjeta.update(HTML(tarjeta_resultado()))

def restablecer(_):
    for p, s in sliders.items():
        s.value = valores_iniciales[p]
    tarjeta.update(HTML(tarjeta_resultado()))

boton_simular.on_click(simular)
boton_reset.on_click(restablecer)

print("=" * 80)
print("SIMULADOR DE ESCENARIOS DE ACTIVIDAD ECONÓMICA")
print("=" * 80)
display(widgets.VBox(list(sliders.values()) + [widgets.HBox([boton_simular, boton_reset])]))

# Una única tarjeta de resultado: arranca con el último mes disponible y se actualiza con cada simulación.
# Queda guardada como salida de la celda, visible aunque el notebook se abra sin ejecutar.
tarjeta = display(HTML(tarjeta_resultado()), display_id=True)
```
